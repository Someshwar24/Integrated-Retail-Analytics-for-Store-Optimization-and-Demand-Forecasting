# **Project Summary -**

The aim of the Integrated Retail Analytics for Store Optimization and Demand Forecasting project is to analyze retail sales data to gain insights into the performance of stores, to spot any anomalies in sales, to cluster stores by their sales' characteristics, and to forecast future demand using machine learning algorithms.

In this project, there is historical sales data at the week level and also store data, markdown data, holidays, and economic data, which include fuel price, CPI, and unemployment. There will be data preprocessing and exploratory data analysis to understand the data set, find anomalies and relationships between variables in the data set.

Some machine learning techniques will be used, which are IQR to detect any anomaly in the data set, K-Means algorithm to cluster the stores, and Random Forest Regression algorithm for demand forecasting. Performance of the forecasting will be measured by MAE and RMSE, whereas the Silhouette Score will measure the performance of the clustering technique.

The last stage is to provide recommendations on how to apply the research conclusions in the context of making decisions related to inventory management, demand forecasting, promotions, and store optimization. The information obtained will allow retailers to make more accurate decisions based on the analysis of the collected data.

```python
# Chart 7: Monthly Sales Trend
df["Month"] = df["Date"].dt.to_period("M")

monthly_sales = df.groupby("Month")["Weekly_Sales"].sum()

monthly_sales.plot(figsize=(12, 5))
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

This chart will help us identify periods with sales peaks and declines. We can then investigate seasonality and possible holiday effects.

```python
# Detect outliers using the IQR method
Q1 = df["Weekly_Sales"].quantile(0.25)
Q3 = df["Weekly_Sales"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df["Sales_Anomaly"] = (
    (df["Weekly_Sales"] < lower) |
    (df["Weekly_Sales"] > upper)
)

print("Total anomalies:", df["Sales_Anomaly"].sum())
print("Total records:", len(df))
```

## ***6. Feature Engineering & Data Pre-processing***

```python
# Check missing values

print(df.isnull().sum()[df.isnull().sum() > 0])

# Fill missing markdown values with 0
markdown_cols = [c for c in df.columns if "MarkDown" in c]

df[markdown_cols] = df[markdown_cols].fillna(0)

# Fill other numeric missing values with median
num_cols = df.select_dtypes(include="number").columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

print("\nRemaining missing values:", df.isnull().sum().sum())
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Week"] = df["Date"].dt.isocalendar().week
df["Day"] = df["Date"].dt.day

# Display new features
display(df[["Date", "Year", "Month", "Week", "Day"]].head())
```

```python
# Store-level sales summary
store_summary = df.groupby("Store").agg(
    avg_sales=("Weekly_Sales", "mean"),
    total_sales=("Weekly_Sales", "sum"),
    sales_std=("Weekly_Sales", "std")
).reset_index()

display(store_summary.head())
```

```python
# Department-level sales summary
dept_summary = df.groupby("Dept").agg(
    avg_sales=("Weekly_Sales", "mean"),
    total_sales=("Weekly_Sales", "sum")
).reset_index()

display(dept_summary.head())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Select features for store segmentation
X = store_summary[["avg_sales", "total_sales", "sales_std"]].fillna(0)

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Create 3 store segments
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
store_summary["Cluster"] = kmeans.fit_predict(X_scaled)

display(store_summary.head())
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_scaled, store_summary["Cluster"])
print("Silhouette Score:", round(score, 2))
```

```python
cluster_summary = store_summary.groupby("Cluster").agg(
    stores=("Store", "count"),
    avg_sales=("avg_sales", "mean"),
    total_sales=("total_sales", "mean")
).sort_values(by="avg_sales",ascending=False).round(2)

display(cluster_summary)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Select features
features = ["Store", "Dept", "Year", "Month", "Week"]
X = df[features]
y = df["Weekly_Sales"]

# Split data into training and testing sets chronologically
df_model = df.dropna(subset=features + ["Weekly_Sales"]).sort_values("Date")
split = int(len(df_model) * 0.8)

X_train = df_model[features].iloc[:split]
X_test = df_model[features].iloc[split:]
y_train = df_model["Weekly_Sales"].iloc[:split]
y_test = df_model["Weekly_Sales"].iloc[split:]

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
# Train Random Forest Regressor
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

# Predict weekly sales
y_pred = model.predict(X_test)

# Evaluate model performance
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2_score = r2_score(y_test,y_pred)

print("Mean Absolute Error (MAE):", round(mae, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
print("r2_score: ", r2_score)
```

```python
# Compare actual and predicted sales
results = pd.DataFrame({
    "Actual_Sales": y_test.values,
    "Predicted_Sales": y_pred
})

results["Error"] = results["Actual_Sales"] - results["Predicted_Sales"]

display(results.head(10).round(2))
```

```python
# Improve Demand Forecasting with External Factors

# Select external features available in the merged dataset
external_cols = [
    "Temperature", "Fuel_Price", "CPI", "Unemployment",
    "MarkDown1", "MarkDown2", "MarkDown3", "MarkDown4", "MarkDown5"
]

features = ["Store", "Dept", "Year", "Month", "Week"] + external_cols

# Prepare model data
df_model = df.dropna(
    subset=["Date", "Weekly_Sales"] + features
).sort_values("Date")

X = df_model[features]
y = df_model["Weekly_Sales"]

# Chronological 80/20 split
split = int(len(df_model) * 0.8)

X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

print("Features used:", features)
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
# Train model with external factors
from sklearn.metrics import r2_score

model_external = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model_external.fit(X_train, y_train)
y_pred_external = model_external.predict(X_test)

mae_external = mean_absolute_error(y_test, y_pred_external)
rmse_external = np.sqrt(mean_squared_error(y_test, y_pred_external))
r2_external = r2_score(y_test,y_pred_external)

print("MAE with external factors:", round(mae_external, 2))
print("RMSE with external factors:", round(rmse_external, 2))
print("r2_score with external factors: ", round(r2_external,2))
```

```python
# Feature importance
importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": model_external.feature_importances_
})

importance = importance.sort_values(
    "Importance", ascending=False
)

display(importance.round(4))
```
